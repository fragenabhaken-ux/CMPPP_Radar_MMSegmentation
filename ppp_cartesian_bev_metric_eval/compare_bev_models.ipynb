```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

output_dir = Path("metric_plots")
output_dir.mkdir(parents=True, exist_ok=True)

for model_name, df_model in df_samples.groupby("model"):
    x = df_model["gt_count"].to_numpy(dtype=float)
    y = df_model["expected_count"].to_numpy(dtype=float)

    valid = np.isfinite(x) & np.isfinite(y)
    x = x[valid]
    y = y[valid]

    fig, ax = plt.subplots(figsize=(6, 6))

    ax.scatter(x, y, alpha=0.5, s=20)

    if len(x) > 0:
        limit_min = min(x.min(), y.min())
        limit_max = max(x.max(), y.max())

        ax.plot(
            [limit_min, limit_max],
            [limit_min, limit_max],
            linestyle="--",
            label=r"Ideal: $\mathbb{E}[N]=N_{\mathrm{GT}}$",
        )

        ax.set_xlim(limit_min, limit_max)
        ax.set_ylim(limit_min, limit_max)

    ax.set_aspect("equal", adjustable="box")
    ax.set_xlabel("Ground-truth count")
    ax.set_ylabel("Expected count")
    ax.set_title(f"Expected count vs. ground-truth count\n{model_name}")
    ax.grid(alpha=0.25)
    ax.legend()
    fig.tight_layout()

    safe_name = str(model_name).replace("/", "_").replace("\\", "_")
    fig.savefig(
        output_dir / f"{safe_name}_expected_count_scatter.png",
        dpi=300,
        bbox_inches="tight",
    )

    plt.show()
```

```python
plot_metrics = [
    "likelihood_optimal_score",
    "radargen_thinning_chamfer_location_m",
    "radargen_thinning_iou_1m",
    "radargen_thinning_da_f1",
]

labels = {
    "likelihood_optimal_score": "Optimal log-likelihood ↑",
    "radargen_thinning_chamfer_location_m": "Location Chamfer [m] ↓",
    "radargen_thinning_iou_1m": "IoU@1m ↑",
    "radargen_thinning_da_f1": "Distance–attribute F1 ↑",
}

for metric in plot_metrics:
    ax = df_samples.boxplot(
        column=metric,
        by="model",
        figsize=(8, 5),
        showfliers=False,
    )
    ax.set_title(labels[metric])
    ax.figure.suptitle("")
    ax.set_xlabel("")
    ax.set_ylabel(labels[metric])
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()
```

```python
metric = "iou_1m"

x_col = f"radargen_thinning_{metric}"
y_col = f"radargen_block_{metric}"

valid = df_samples[[x_col, y_col]].notna().all(axis=1)
data = df_samples.loc[valid]

plt.figure(figsize=(6, 6))
plt.scatter(data[x_col], data[y_col], alpha=0.4, s=18)

low = min(data[x_col].min(), data[y_col].min())
high = max(data[x_col].max(), data[y_col].max())
plt.plot([low, high], [low, high], linestyle="--")

plt.xlabel("Thinning IoU@1m")
plt.ylabel("Block IoU@1m")
plt.title("Thinning versus block-sampling IoU@1m")
plt.gca().set_aspect("equal", adjustable="box")
plt.tight_layout()
plt.show()
```

This is expected from the block approximation. It activates a block with probability 1−exp(−m_block) and places at most one point inside each active block. It therefore suppresses multiple nearby points and changes the local point density, which directly reduces the tolerance-based IoU. The thinning sampler instead produces an exact continuous PPP sample from the raster intensity. The implementation consequently reports both sampling methods separately rather than averaging them.
