```python
import pandas as pd

ppp_metrics = [
    "count_rmse",
    "likelihood_optimal_score_mean",
    "velocity_mae_global",
    "velocity_nonzero_mae_global",
    "rcs_mae_global",
]

ppp_metric_labels = {
    "count_rmse": "Count RMSE",
    "likelihood_optimal_score_mean": "Optimal logL score",
    "velocity_mae_global": "Velocity MAE",
    "velocity_nonzero_mae_global": "Non-zero velocity MAE",
    "rcs_mae_global": "RCS MAE",
}


def format_estimate_with_ci(
    estimate: float,
    ci_low: float,
    ci_high: float,
    decimals: int = 3,
) -> str:
    if pd.isna(estimate):
        return "—"

    if pd.isna(ci_low) or pd.isna(ci_high):
        return f"{estimate:.{decimals}f}"

    return (
        f"{estimate:.{decimals}f} "
        f"[{ci_low:.{decimals}f}, {ci_high:.{decimals}f}]"
    )


ppp_table = df_summary[
    [
        "model",
        *ppp_metrics,
        *[
            column
            for metric in ppp_metrics
            for column in (
                f"{metric}_ci95_low",
                f"{metric}_ci95_high",
            )
        ],
    ]
].copy()

# Sort before replacing numeric values by formatted strings.
ppp_table = ppp_table.sort_values(
    by="count_rmse",
    ascending=True,
).reset_index(drop=True)

for metric in ppp_metrics:
    ppp_table[metric] = ppp_table.apply(
        lambda row: format_estimate_with_ci(
            estimate=row[metric],
            ci_low=row[f"{metric}_ci95_low"],
            ci_high=row[f"{metric}_ci95_high"],
            decimals=3,
        ),
        axis=1,
    )

# Remove the now-redundant separate CI columns.
ppp_table = ppp_table[
    ["model", *ppp_metrics]
].rename(
    columns={
        "model": "Model",
        **ppp_metric_labels,
    }
)

ppp_table
```

```python
metric = "iou_1m"

x_col = f"radargen_thinning_{metric}"
y_col = f"radargen_block_{metric}"

valid = df_samples[[x_col, y_col]].notna().all(axis=1)
data = df_samples.loc[valid]

plt.figure(figsize=(6, 6))
plt.scatter(data[x_col], data[y_col], alpha=0.4, s=18)

low = min(data[x_col].min(), data[y_col].min())
high = max(data[x_col].max(), data[y_col].max())
plt.plot([low, high], [low, high], linestyle="--")

plt.xlabel("Thinning IoU@1m")
plt.ylabel("Block IoU@1m")
plt.title("Thinning versus block-sampling IoU@1m")
plt.gca().set_aspect("equal", adjustable="box")
plt.tight_layout()
plt.show()
```

This is expected from the block approximation. It activates a block with probability 1−exp(−m_block) and places at most one point inside each active block. It therefore suppresses multiple nearby points and changes the local point density, which directly reduces the tolerance-based IoU. The thinning sampler instead produces an exact continuous PPP sample from the raster intensity. The implementation consequently reports both sampling methods separately rather than averaging them.
